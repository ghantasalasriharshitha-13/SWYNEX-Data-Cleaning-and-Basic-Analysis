```python
import gzip
import pandas as pd
import io

compressed_path = '/content/uniprotkb_accession_P01308_2026_10_19.xlsx.gz'

# Decompress the gzipped excel file in memory and load it with pandas
with gzip.open(compressed_path, 'rb') as f:
    decompressed_data = io.BytesIO(f.read())

df = pd.read_excel(decompressed_data, engine='openpyxl')

# Display basic information and first few rows
print(f"Dataset Shape: {df.shape}")
display(df.head())
```
